## 2. The 5 Matrix Operations

Same examples as the video — let NumPy verify our hand-calculations.

### 2.1 Addition & Subtraction

**Rule:** Shapes must match. It's element-wise.

$$A = \begin{bmatrix} 2 & 3 \\ 4 & 1 \end{bmatrix}, \quad B = \begin{bmatrix} 1 & 0 \\ 2 & 5 \end{bmatrix}$$

```python
A = np.array([
    [2, 3],
    [4, 1]
])

B = np.array([
    [1, 0],
    [2, 5]
])

sum = A + B
difference = A - B

print("A + B =\n", sum)
print("\nA - B =\n", difference)
```

## 2.2 Transpose — flip rows ↔ columns

```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])

A_T = A.T

print("A shape  :", A.shape)    # (2, 3)
print("A:\n", A)

print("\nA.T shape:", A_T.shape)  # (3, 2)
print("A.T:\n", A_T)
```

### 2.3 Scalar Multiplication

$$\alpha = 0.01, \quad W = \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}$$

```python
alpha = 0.01
W = np.array([[1, 2],
              [3, 4]])

print("alpha * W =\n", alpha * W)
```

### 2.4 Element-wise Multiplication (denoted `⊙`)

Multiply matching positions. Different from matrix multiplication!

```python
A = np.array([[2, 3],
              [4, 1]])

B = np.array([[1, 0],
              [2, 5]])

# In NumPy, the * operator on arrays is element-wise (NOT matrix mult)
print("A ⊙ B =\n", A * B)
```

### 2.5 Matrix Multiplication — the big one

**Rule:** inner dims must match. `(m × n) · (n × p) = (m × p)`.

Same example from the video:

$$A = \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}, \quad B = \begin{bmatrix} 5 & 6 \\ 7 & 8 \end{bmatrix}$$

```python
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

# Use @ for matrix multiplication (or np.matmul / np.dot)
result = A @ B

print("A @ B =\n", result)
print("shape:", result.shape)
```

Expected from the video:
$$\begin{bmatrix} 19 & 22 \\ 43 & 50 \end{bmatrix}$$
Matches ✓

Let's manually verify just one cell — top-left `(1×5) + (2×7) = 5 + 14 = 19` ✓

## 3. Putting it together — A Single Neuron's Forward Pass

From the video: a single neuron computes

$$z = w \cdot x + b \quad\rightarrow\quad \hat{y} = \sigma(z)$$

Let's build it. Inputs: 3 features. Weights: 3 values. One bias. One sigmoid activation.

```python
# inputs
x = np.array([0.5, 0.2, 0.9])   # 3 input features

# parameters
w = np.array([0.4, -0.3, 0.8])  # 3 weights (one per feature)
b = 0.1

print("x shape:", x.shape)
print("w shape:", w.shape)


# step 1: Linear combination: z = w.x + b
z = w @ x + b
print("\nz=", z)

# step 2: activation y_hat = sigmoid(x)
def sigmoid(value):
  return 1 / (1 + np.exp(-value))

y_hat = sigmoid(z)
print("\ny_hat=", y_hat)
```

## 4. Scaling up — A Hidden Layer with 4 Neurons

Now instead of 1 neuron, we have a layer of 4 neurons all looking at the same input. The weights become a *matrix*.

$$Z = W \cdot x + b$$

- `x` : shape `(3,)`  — 3 input features
- `W` : shape `(4, 3)` — 4 neurons, each with 3 weights
- `b` : shape `(4,)`  — one bias per neuron
- `Z` : shape `(4,)`  — output of the layer

```python
x = np.array([.5, 0.2, 00.9])           # 3 features

W = np.array([                            # 4 neurons × 3 weights
    [ 0.4, -0.3,  0.8],
    [ 0.1,  0.6, -0.2],
    [-0.5,  0.2,  0.4],
    [ 0.7,  0.0, -0.1]
])

b = np.array([0.1, -0.1, 0.0, 0.2])      # one bias per neuron

# Forward pass — entire layer in ONE matrix multiplication
Z = W @ x + b
A = sigmoid(Z)

print("Z (pre-activation):", Z)
print("A (post-activation):", A)
print("\nOutput shape:", A.shape, "-> 4 neurons, 4 activations")
```

That single line `Z = W @ x + b` is what every deep learning framework (PyTorch, TensorFlow) runs millions of times per second under the hood. Now you know what it actually does.

## 5. Recap — Where Each Operation Shows Up in DL

| Operation        | Where it's used                              |
|------------------|----------------------------------------------|
| `W @ x`          | forward pass                                  |
| `W @ x + b`      | adding bias                                   |
| `α · ∇w`         | applying learning rate (scalar mult)          |
| `a ⊙ b`          | dropout, gating (element-wise mult)           |
| `W.T`            | backpropagation through a linear layer        |
