## 📚 Phase 2: Document Pipeline in Action

Let's demonstrate our advanced document processing pipeline that transforms raw LangChain documentation into searchable, semantically meaningful chunks.

```python
print("📚 DOCUMENT PROCESSING PIPELINE DEMONSTRATION")
print("=" * 50)

pipeline_demo_start = time.time()

try:
    # Initialize document processing components
    print("\n🔧 Initializing Document Processing Components...")
    
    # Repository manager for LangChain documentation
    repo_manager = LangChainRepoManager()
    
    # Check if repository exists
    repo_path = project_root / "data" / "langchain_repo"
    repo_exists = repo_path.exists()
    
    print(f"   📁 Repository path: {repo_path}")
    print(f"   📋 Repository exists: {repo_exists}")
    
    if repo_exists:
        # Get documentation structure
        print("\n📖 Analyzing Documentation Structure...")
        doc_paths = repo_manager.get_documentation_paths()
        
        total_files = 0
        structure_info = {}
        
        for category, file_list in doc_paths.items():
            file_count = len(file_list)
            total_files += file_count
            structure_info[category] = file_count
            print(f"   📂 {category}: {file_count} files")
        
        print(f"\n   📊 Total documentation files: {total_files}")
        demo_results['demo_outputs']['doc_structure'] = structure_info
        
        # Demonstrate document parsing
        print("\n🔍 Document Parsing Demonstration...")
        
        from ingestion.document_parser import MultiFormatParser, DocumentMetadata
        
        parser = MultiFormatParser()
        parsing_examples = []
        
        # Sample a few files for demonstration
        sample_files = []
        for category, files in doc_paths.items():
            if files:
                sample_files.extend(files[:2])  # Take first 2 from each category
        
        for i, file_path in enumerate(sample_files[:5]):  # Limit to 5 examples
            try:
                parse_start = time.time()
                parsed_doc = parser.parse_document(file_path)
                parse_time = (time.time() - parse_start) * 1000
                
                if parsed_doc:
                    content_length = len(parsed_doc['content'])
                    metadata = parsed_doc['metadata']
                    
                    parsing_examples.append({
                        'file_name': file_path.name,
                        'file_type': metadata.get('file_extension', 'unknown'),
                        'content_length': content_length,
                        'parse_time_ms': parse_time,
                        'doc_type': metadata.get('doc_type', 'unknown')
                    })
                    
                    print(f"   ✅ {file_path.name} ({metadata.get('file_extension', 'unknown')}): {content_length:,} chars in {parse_time:.1f}ms")
                else:
                    print(f"   ⚠️ {file_path.name}: Parse failed")
                    
            except Exception as e:
                print(f"   ❌ {file_path.name}: {str(e)}")
        
        demo_results['demo_outputs']['parsing_examples'] = parsing_examples
        
        if parsing_examples:
            avg_parse_time = sum(ex['parse_time_ms'] for ex in parsing_examples) / len(parsing_examples)
            total_content = sum(ex['content_length'] for ex in parsing_examples)
            print(f"\n   📊 Parsing Performance:")
            print(f"      Average parse time: {avg_parse_time:.1f}ms")
            print(f"      Total content processed: {total_content:,} characters")
            print(f"      Processing rate: {total_content/max(1,sum(ex['parse_time_ms'] for ex in parsing_examples)*0.001):,.0f} chars/second")
        
        # Demonstrate semantic chunking
        print("\n✂️ Semantic Chunking Demonstration...")
        
        from chunking.text_splitter import SemanticHierarchicalChunker
        
        chunker = SemanticHierarchicalChunker(
            chunk_size=settings.chunk_size,
            chunk_overlap=settings.chunk_overlap
        )
        
        if parsing_examples:
            # Use the largest parsed document for chunking demo
            largest_file = max(parsing_examples, key=lambda x: x['content_length'])
            
            # Re-parse the largest file to get content
            largest_file_path = None
            for file_path in sample_files:
                if file_path.name == largest_file['file_name']:
                    largest_file_path = file_path
                    break
            
            if largest_file_path:
                chunking_start = time.time()
                parsed_doc = parser.parse_document(largest_file_path)
                
                if parsed_doc:
                    chunks = chunker.split_text(parsed_doc['content'], parsed_doc['metadata'])
                    chunking_time = (time.time() - chunking_start) * 1000
                    
                    chunk_sizes = [len(chunk['content']) for chunk in chunks]
                    avg_chunk_size = sum(chunk_sizes) / len(chunk_sizes) if chunks else 0
                    
                    print(f"   📄 Document: {largest_file['file_name']}")
                    print(f"   📏 Original length: {largest_file['content_length']:,} characters")
                    print(f"   ✂️ Created chunks: {len(chunks)}")
                    print(f"   📊 Average chunk size: {avg_chunk_size:.0f} characters")
                    print(f"   ⏱️ Chunking time: {chunking_time:.1f}ms")
                    
                    # Show chunk distribution
                    if len(chunks) > 0:
                        min_chunk = min(chunk_sizes)
                        max_chunk = max(chunk_sizes)
                        print(f"   📈 Chunk size range: {min_chunk} - {max_chunk} characters")
                        
                        # Show first chunk preview
                        if chunks[0]['content']:
                            preview = chunks[0]['content'][:200] + "..." if len(chunks[0]['content']) > 200 else chunks[0]['content']
                            print(f"   📖 First chunk preview: {preview}")
                    
                    demo_results['demo_outputs']['chunking_demo'] = {
                        'document_name': largest_file['file_name'],
                        'original_length': largest_file['content_length'],
                        'chunks_created': len(chunks),
                        'avg_chunk_size': avg_chunk_size,
                        'chunking_time_ms': chunking_time,
                        'chunk_size_range': [min_chunk, max_chunk] if chunks else [0, 0]
                    }
        
        # Demonstrate embedding generation (conceptual - not running actual generation)
        print("\n🧠 Embedding Generation Overview...")
        print(f"   🎯 Target embedding model: {settings.embedding_model}")
        print(f"   📐 Embedding dimensions: 768 (text-embedding-004)")
        print(f"   ⚙️ Processing configuration:")
        print(f"      Batch size: 100 chunks")
        print(f"      Caching enabled: Yes")
        print(f"      Retry logic: 3 attempts")
        print(f"   🗄️ Vector storage: Qdrant Cloud")
        print(f"   📊 Collection: {settings.collection_name}")
        
    else:
        print("\n⚠️ LangChain repository not found")
        print("   To see the full document processing demo:")
        print("   1. Run the data ingestion notebook first")
        print("   2. Or manually clone the LangChain repository")
        
        demo_results['demo_outputs']['doc_processing_status'] = 'repository_not_found'

except Exception as e:
    print(f"❌ Document processing demo error: {str(e)}")
    demo_results['demo_outputs']['doc_processing_error'] = str(e)

pipeline_demo_time = (time.time() - pipeline_demo_start) * 1000
print(f"\n✅ Document pipeline demonstration completed in {pipeline_demo_time:.1f}ms")

# Summary of document processing capabilities
print(f"\n🎯 DOCUMENT PROCESSING CAPABILITIES SUMMARY")
print("-" * 45)
processing_capabilities = [
    "✅ Multi-format parsing (MD, MDX, RST, IPYNB, PY)",
    "✅ Metadata extraction and enrichment",
    "✅ Semantic-aware hierarchical chunking",
    "✅ Code block preservation and formatting",
    "✅ Efficient batch processing with error handling",
    "✅ Configurable chunk sizes and overlap",
    "✅ Document type classification and tagging",
    "✅ Content deduplication and filtering"
]

for capability in processing_capabilities:
    print(f"   {capability}")

demo_results['performance_metrics']['document_pipeline_time'] = pipeline_demo_time
```

## 🔍 Phase 3: Advanced Search & Retrieval Showcase

Experience our sophisticated search capabilities including semantic search, multi-query fusion, and hybrid approaches.

```python
print("🔍 ADVANCED SEARCH & RETRIEVAL SHOWCASE")
print("=" * 45)

search_demo_start = time.time()
search_results = {}

try:
    # Initialize search components
    print("\n⚙️ Initializing Search Components...")
    
    # Semantic searcher
    semantic_searcher = create_semantic_searcher()
    print(f"   ✅ Semantic searcher initialized")
    
    # Hybrid searcher
    hybrid_searcher = create_hybrid_searcher()
    print(f"   ✅ Hybrid searcher initialized")
    
    # Demo queries that showcase different search capabilities
    demo_queries = [
        {
            'query': 'How to create custom LangChain agents with memory?',
            'category': 'Technical Implementation',
            'complexity': 'Advanced'
        },
        {
            'query': 'What are the different types of document loaders?',
            'category': 'Conceptual Overview',
            'complexity': 'Intermediate'
        },
        {
            'query': 'LangChain streaming callbacks implementation',
            'category': 'Technical Keywords',
            'complexity': 'Advanced'
        },
        {
            'query': 'Best practices for prompt engineering',
            'category': 'Best Practices',
            'complexity': 'Intermediate'
        }
    ]
    
    print(f"\n📝 Demo queries prepared: {len(demo_queries)}")
    
    # Demonstrate different search methods
    search_methods = {
        'semantic': 'Semantic Search (Embeddings)',
        'hybrid': 'Hybrid Search (Semantic + Keywords)',
        'multi_query': 'Multi-Query Fusion'
    }
    
    for method_key, method_name in search_methods.items():
        print(f"\n{method_name.upper()}")
        print("-" * len(method_name))
        
        method_results = []
        
        for query_info in demo_queries[:2]:  # Test first 2 queries per method
            query = query_info['query']
            print(f"\n🔍 Query: {query}")
            print(f"   Category: {query_info['category']} | Complexity: {query_info['complexity']}")
            
            try:
                search_start = time.time()
                
                if method_key == 'semantic':
                    results = semantic_searcher.search(query, top_k=5)
                elif method_key == 'hybrid':
                    results = hybrid_searcher.hybrid_search(query, top_k=5, rerank=True)
                elif method_key == 'multi_query':
                    # Generate related queries for multi-query demo
                    related_queries = [
                        query,
                        f"Examples of {query.split()[-1] if query.split() else query}",
                        f"Tutorial for {' '.join(query.split()[:3])}" if len(query.split()) >= 3 else query
                    ]
                    results = semantic_searcher.multi_query_search(related_queries, top_k=5, fusion_method="rrf")
                
                search_time = (time.time() - search_start) * 1000
                
                if results:
                    # Calculate search quality metrics
                    scores = [r.get('score', 0) for r in results]
                    avg_score = sum(scores) / len(scores)
                    top_score = max(scores) if scores else 0
                    
                    # Analyze document types in results
                    doc_types = [r.get('metadata', {}).get('doc_type', 'unknown') for r in results]
                    unique_doc_types = len(set(doc_types))
                    
                    result_info = {
                        'query': query,
                        'method': method_key,
                        'results_count': len(results),
                        'search_time_ms': search_time,
                        'avg_score': avg_score,
                        'top_score': top_score,
                        'doc_type_diversity': unique_doc_types
                    }
                    
                    method_results.append(result_info)
                    
                    print(f"   ✅ Results: {len(results)} documents")
                    print(f"   ⏱️ Search time: {search_time:.1f}ms")
                    print(f"   📊 Average score: {avg_score:.3f}")
                    print(f"   🏆 Top score: {top_score:.3f}")
                    print(f"   📂 Document types: {unique_doc_types} unique types")
                    
                    # Show top result details
                    if results and DEMO_CONFIG['show_detailed_outputs']:
                        top_result = results[0]
                        metadata = top_result.get('metadata', {})
                        content_preview = top_result.get('content', '')[:150] + "..."
                        
                        print(f"\n   🔍 Top Result Details:")
                        print(f"      File: {metadata.get('file_name', 'Unknown')}")
                        print(f"      Type: {metadata.get('doc_type', 'Unknown')}")
                        print(f"      Score: {top_result.get('score', 0):.3f}")
                        print(f"      Preview: {content_preview}")
                        
                        # Show hybrid-specific metadata if available
                        if method_key == 'hybrid' and 'hybrid_metadata' in top_result:
                            hybrid_meta = top_result['hybrid_metadata']
                            print(f"      Semantic contrib: {hybrid_meta.get('semantic_contribution', 0):.3f}")
                            print(f"      Keyword contrib: {hybrid_meta.get('keyword_contribution', 0):.3f}")
                            print(f"      Fusion method: {hybrid_meta.get('fusion_method', 'unknown')}")
                
                else:
                    print(f"   ⚠️ No results found")
                    method_results.append({
                        'query': query,
                        'method': method_key,
                        'results_count': 0,
                        'search_time_ms': search_time,
                        'avg_score': 0,
                        'top_score': 0,
                        'doc_type_diversity': 0
                    })
                
            except Exception as e:
                print(f"   ❌ Search failed: {str(e)}")
                method_results.append({
                    'query': query,
                    'method': method_key,
                    'error': str(e),
                    'results_count': 0,
                    'search_time_ms': 0
                })
        
        search_results[method_key] = method_results
        
        # Method performance summary
        if method_results:
            successful_searches = [r for r in method_results if r.get('results_count', 0) > 0]
            if successful_searches:
                avg_time = sum(r['search_time_ms'] for r in successful_searches) / len(successful_searches)
                avg_results = sum(r['results_count'] for r in successful_searches) / len(successful_searches)
                avg_quality = sum(r['avg_score'] for r in successful_searches) / len(successful_searches)
                
                print(f"\n   📊 {method_name} Performance Summary:")
                print(f"      Success rate: {len(successful_searches)}/{len(method_results)} ({len(successful_searches)/len(method_results)*100:.1f}%)")
                print(f"      Avg search time: {avg_time:.1f}ms")
                print(f"      Avg results per query: {avg_results:.1f}")
                print(f"      Avg result quality: {avg_quality:.3f}")
    
    demo_results['demo_outputs']['search_results'] = search_results
    
    # Search comparison analysis
    print(f"\n📊 SEARCH METHOD COMPARISON")
    print("-" * 35)
    
    comparison_data = []
    for method, results in search_results.items():
        successful = [r for r in results if r.get('results_count', 0) > 0]
        if successful:
            comparison_data.append({
                'method': method,
                'avg_time': sum(r['search_time_ms'] for r in successful) / len(successful),
                'avg_quality': sum(r['avg_score'] for r in successful) / len(successful),
                'success_rate': len(successful) / len(results) * 100,
                'avg_results': sum(r['results_count'] for r in successful) / len(successful)
            })
    
    if comparison_data:
        # Find best performing method
        best_quality = max(comparison_data, key=lambda x: x['avg_quality'])
        fastest = min(comparison_data, key=lambda x: x['avg_time'])
        most_results = max(comparison_data, key=lambda x: x['avg_results'])
        
        print(f"   🏆 Best Quality: {best_quality['method'].title()} (score: {best_quality['avg_quality']:.3f})")
        print(f"   ⚡ Fastest: {fastest['method'].title()} ({fastest['avg_time']:.1f}ms avg)")
        print(f"   📊 Most Results: {most_results['method'].title()} ({most_results['avg_results']:.1f} avg)")
        
        demo_results['comparisons']['search_methods'] = comparison_data
    
    # Advanced search features demonstration
    print(f"\n🚀 ADVANCED SEARCH FEATURES")
    print("-" * 30)
    
    advanced_features = [
        "✅ Reciprocal Rank Fusion (RRF) for multi-query optimization",
        "✅ Configurable semantic/keyword weight balancing", 
        "✅ Real-time re-ranking with cross-encoder techniques",
        "✅ Document type filtering and preference boosting",
        "✅ Similarity threshold filtering with adaptive scoring",
        "✅ Batch processing with parallel query execution",
        "✅ Context-aware search with conversation history",
        "✅ Module-specific search with namespace filtering"
    ]
    
    for feature in advanced_features:
        print(f"   {feature}")

except Exception as e:
    print(f"❌ Search demonstration error: {str(e)}")
    demo_results['demo_outputs']['search_error'] = str(e)

search_demo_time = (time.time() - search_demo_start) * 1000
print(f"\n✅ Search & retrieval showcase completed in {search_demo_time:.1f}ms")
demo_results['performance_metrics']['search_demo_time'] = search_demo_time
```
